```python
import numpy as np
customers = np.array([
 [22, 2, 15, 1],
 [25, 4, 20, 2],
 [47, 10, 120, 12],
 [52, 12, 150, 15],
 [35, 6, 60, 6],
 [38, 7, 70, 7],
 [19, 1, 10, 1],
 [60, 8, 130, 10],
 [41, 9, 90, 9],
 [28, 5, 35, 4],
 [33, 6, 55, 5],
 [55, 11, 160, 14]
])

print(customers.dtype)
print(customers.size)
print(customers.shape)
print(customers.ndim)
```

```python
minn = customers.min(axis=0)
print("min: ",minn)
maxx = customers.max(axis=0)
print("max: ",maxx)
mean = customers.mean(axis=0)
print("mean: ",mean)
std = customers.std(axis=0)
print("std: ",std)
```

```python

feature_names = np.array([
 "age",
 "monthly_visits",
 "avg_order_value",
 "total_purchases"
])

for name, mn, mx, avg, sd in zip(feature_names, minn, maxx, mean, std):
    print(f"features: {name}")
    print(f"min: {mn}")
    print(f"max: {mx}")
    print(f"mean: {avg}")
    print(f"std: {std}")
    
print(np.min(customers[:,0]))
```

```python
f_mean = np.argmax(mean)
print("max feature: ",f_mean)
print("mean: ", mean[f_mean])
```

```python
std_max = np.argmax(std)
print("std: ",std[std_max])
```

```python
customers_scaled = (customers - minn) / (maxx - minn)
print(customers_scaled)
```

```python
print("min: ",np.min(customers_scaled))
print("max: ", np.max(customers_scaled))
```

```python
mean_scaled = np.mean(customers_scaled,axis=0)
print(mean_scaled)

```

```python
#section 4
ref_points = np.array([
 [23, 2, 15, 1], # low activity
 [32, 5, 45, 4], # normal
 [42, 8, 85, 8], # loyal
 [55, 12, 150, 14] # VIP
])

ref_names = np.array([
 "low activity",
 "normal",
 "loyal",
 "VIP"
])

ref_point_scaled = ((ref_points - minn) / (maxx - minn))
print(ref_point_scaled)
```

```python
diff = (customers_scaled[:,np.newaxis, :]) - ref_point_scaled[np.newaxis,:, :]
distances = np.linalg.norm(diff, axis=2)
print(distances.shape)
```

```python
nearest = np.argmin(distances, axis=1)
print(nearest)
for i, index in enumerate(nearest):
    print(f"Customer {i + 1} -> {ref_names[index]}")
```

```python
feature_names = [
    "age",
    "monthly visits",
    "avg order value",
    "total purchases"
]

cluster_means = []

for i, cluster_name in enumerate(ref_names):
    members = customers[nearest == i]
    mean_features = members.mean(axis=0)
    cluster_means.append(mean_features)

    print(f"Cluster: {cluster_name}")
    print(f"Count: {len(members)}")

    for feature, mean_value in zip(feature_names, mean_features):
        print(f"Mean {feature}: {mean_value:}")

    print()

cluster_means = np.array(cluster_means)

highest_order_cluster = np.argmax(cluster_means[:, 2])
highest_visits_cluster = np.argmax(cluster_means[:, 1])

print("Highest avg order value:", ref_names[highest_order_cluster])
print("Highest monthly visits:", ref_names[highest_visits_cluster])
```

```python
#6
total_purchases = customers[:, 3]
avg_order_value = customers[:, 2]
monthly_visits = customers[:, 1]

label_customer = np.where(total_purchases >= 8, "valuable", "regular")
print(label_customer)

valuable = label_customer == "valuable"
regular = label_customer == "regular"

print("Valuable count:", np.sum(valuable))
print("Regular count:", np.sum(regular))

print("Valuable mean avg order value:", avg_order_value[valuable].mean())
print("Regular mean avg order value:", avg_order_value[regular].mean())

print("Valuable mean monthly visits:", monthly_visits[valuable].mean())
print("Regular mean monthly visits:", monthly_visits[regular].mean())
```

```python
#7
outlier_customer = np.array([[30, 1, 500, 1]])
customers_with_outlier = np.concatenate((customers,outlier_customer),axis=0)
print(customers_with_outlier.shape)
```

```python
min_outlaw = customers_with_outlier.min()
max_outlaw = customers_with_outlier.max()
outlaw_scaled = (customers_with_outlier - min_outlaw) / (max_outlaw - min_outlaw)
print(outlaw_scaled)
```

```python

center = np.mean(outlaw_scaled, axis=0)
distancee = np.linalg.norm((center-outlaw_scaled),axis=1)
print(distancee)
max_distance = np.argmax(distancee)
print(max_distance)
print(customers_with_outlier[max_distance])
```
