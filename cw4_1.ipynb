```python
import numpy as np
# study_hours, attendance_percent, previous_score, final_score
students = np.array([
 [2, 60, 45, 50],
 [5, 80, 70, 75],
 [1, 40, 35, 38],
 [8, 90, 88, 92],
 [4, 75, 60, 65],
 [7, 85, 80, 84],
 [3, 55, 50, 52],
 [6, 70, 72, 78]
])
dtype = students.dtype
print("dtype: " ,dtype)
size = students.size
print("size: " ,size)
shape = students.shape
print("shape: " ,  shape)
ndim = students.ndim
print("ndim: " , ndim)
print("sample: ",len(students))
features = len(students[0,:])
print("features: ", features)
x = students[:, :3]
print("x: ", x)
y = students[:, -1]
print("y: ", y)
print("x shape: ", x.shape)
print("y shape: ", y.shape)
```

```python
#2
rule_based_label = np.where( y>60, "pass", "fail")
print(rule_based_label)
print(np.sum(rule_based_label == "pass"))
print(np.sum(rule_based_label == "fail"))


```

```python
#3
study_hours = x[:, 0].mean()
print("mean_study_hours: ", study_hours)
max_attendance = x[:, 1].max()
print("max_attendance: ", max_attendance)
min_attendance = x[:, 1].min()
print("min_attendance: ", min_attendance)
attendance_percent = x[:, 1]
effort_score = study_hours * attendance_percent / 100
print("effort_score: ", effort_score)
X_new = np.column_stack((x, effort_score))
print("X_new: ", X_new.shape)
```

```python
#4
class_label = np.where(y>60 ,"pass","fail")
print(class_label)
passed_percentage = np.sum(rule_based_label == "pass")/len(y)*100
print("passed_percentage: ",passed_percentage)
```

```python
#5
label_1 = np.where(y>=80, "strong", y)
label_2 = np.where((y>=60) & (y<80), "normal", label_1)
label_3 = np.where(y<60, "weak", label_2)
print("label_1: ", label_1)
print("label_2: ", label_2)
print("label_3: ", label_3)
print("weak students: ", np.sum(label_3 == "weak"))
print("normal students: ", np.sum(label_3 == "normal"))
print("strong students: ", np.sum(label_3 == "strong"))
weak_mean = X_new[label_3 == "weak", 0].mean()
print("weak: ",weak_mean)
normal_mean = X_new[label_3 == "normal", 0].mean()
print("normal: ",normal_mean)
strong_mean = X_new[label_3 == "strong", 0].mean()
print("strong: ",strong_mean)
# print(f"Weak students average study hours: {np.mean((X_new[np.where(label_3=="weak"),0]))}")
# print(f"Normal students average study hours: {np.mean((X_new[np.where(label_3=="normal"),0]))}")
# print(f"Strong students average study hours: {np.mean((X_new[np.where(label_3=="strong"),0]))}")
```
