```python
import numpy as np
students = np.array([
 [2, 60, 45, 50],
 [5, 80, 70, 75],
 [1, 40, 35, 38],
 [8, 90, 88, 92],
 [4, 75, 60, 65],
 [7, 85, 80, 84],
 [3, 55, 50, 52],
 [6, 70, 72, 78]
])
print(students.shape)
```

```python
x = students[:,0:3]
print(x)
```

```python
y = students[:,-1]
print(y)
```

```python
rule_based_label = np.where(y>60, "pass", "fail")
print(rule_based_label)
print(np.sum(rule_based_label == "pass"))
print(len(np.where(rule_based_label == "fail")[0]))
```

```python
students_mean = np.mean(students[:,0])
print("students_mean: ",students_mean)
```

```python
max__attendance = np.min(students[:,1])
min__attendance = np.max(students[:,1])
print("max: ",max__attendance)
print("min: ",min__attendance)
```

```python
study_hours = x[:,0]
attendance_percent = x[:,1]
effort_score = study_hours * attendance_percent / 100
# new_x = np.column_stack((x,effort_score))
# print(new_x)
new_x = np.concatenate((x, effort_score.reshape(-1, 1)), axis=1)
print(new_x)
print(effort_score.shape)
print(new_x.shape)
```

```python
final_score = y
class_label = np.where(final_score>60, "pass","fail")
print(class_label)
passed = np.sum(class_label == "pass")
failed = np.sum(class_label == "fail")
print("passed: ",passed,"\n","failed: ",failed)
percentage = (passed/len(class_label))*100
print(percentage)
```

```python
level_label = np.where(final_score < 60 ,"weak",
                       np.where(final_score >= 80,"strong","normal"))
print(level_label)
```

```python
print(np.sum(level_label == "strong"))
print(np.sum(level_label == "normal"))
print(np.sum(level_label == "weak"))
```

```python
print(x)
# study_hours, attendance_percent, previous_score
study_level = np.column_stack((study_hours,level_label))
print(study_level)
```

```python
scores = study_level[:, 0].astype(float)
labels = study_level[:, 1]

for label in ["weak", "normal", "strong"]:
    mean_score = scores[labels == label].mean()
    print(label, mean_score)
```

```python
mean_x = x.mean(axis=0)
min_x = x.min(axis=0)
max_x = x.max(axis=0)
std_x = x.std(axis=0)
print("mean:",mean_x,"min:",min_x,"max:",max_x,"std:",std_x)
```

```python
x_scaled = (x - min_x) / (max_x - min_x)
print(x_scaled)
print(x_scaled.max())
print(x_scaled.min())
```

```python
new_student = np.array([5, 75, 68])
new_student_scaled = (new_student - min_x) / (max_x - min_x)
print(new_student_scaled)
euclidean = np.linalg.norm(new_student_scaled-x_scaled,axis=1)
print(euclidean)
closest = np.argmin(euclidean)
print(x_scaled[closest])
print(y[closest])
print(level_label[closest])
```

```python
new_students = np.array([
 [2, 50, 40],
 [6, 82, 78],
 [4, 65, 58],
 [8, 88, 85]
])
new_students_scaled = (new_students - min_x) / (max_x - min_x)
diff = new_students_scaled[:, None, :] - x_scaled[None, :, :]
distance = np.linalg.norm((diff),axis=2)
print(distance.shape)
nearest_indices = np.argmin(distance, axis=1)
predicted_labels = labels[nearest_indices]

for i, label in enumerate(predicted_labels, start=1):
    print(f"Student {i} predicted label: {label}")
```
