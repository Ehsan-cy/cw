```python
#5
patient_id = np.array([1, 2, 3, 4, 5])
ground_truth = np.array([
 "Healthy",
 "Malignant",
 "Healthy",
 "Healthy",
 "Malignant"
])

predictions = np.array([
 "Healthy",
 "Malignant",
 "Malignant",
 "Healthy",
 "Healthy"
])
predicct = ground_truth == predictions
print(predicct)
corrects = np.sum(ground_truth == predictions)
print(corrects)
incorrects = np.sum(ground_truth != predictions)
print(incorrects)

Accuracy = (corrects / len(ground_truth))*100
print("accuracy: ",Accuracy)
error = (incorrects / len(ground_truth)*100)
print("error: ",error)

false_positive = np.where(( ground_truth == "Healthy" ) & ( predictions == "Malignant" ), 1 , 0 )
print(np.sum(false_positive))

false_negative= np.sum(
(predictions == "Healthy") &
(ground_truth == "Malignant")
)
print(false_negative)

FP_mask = ((ground_truth == "Healthy") & (predictions == "Malignant"))
FN_mask = ((ground_truth == "Malignant") & (predictions == "Healthy"))
FP = np.sum(FP_mask)
FN = np.sum(FN_mask)
print("FP patients:", patient_id[FP_mask])
print("FN patients:", patient_id[FN_mask])
print("FP_mask:",FP_mask)
print("FN_mask:",FN_mask)
 
```

```python
email_id = np.array([1, 2, 3, 4, 5])
ground_truth = np.array([
 "Not Spam",
 "Spam",
 "Not Spam",
 "Not Spam",
 "Spam"
])
predictions = np.array([
 "Not Spam",
 "Spam",
 "Spam",
 "Not Spam",
 "Not Spam"
])
predicts = ground_truth == predictions
print(predicts)
corrects = np.sum(ground_truth == predictions)
incorrects = np.sum(ground_truth != predictions)
Accuracy = (corrects / len(ground_truth)*100)
error = (incorrects / len(ground_truth)*100)
print("accuracy: ",Accuracy)
print("error: ",error)

FP_mask = ((ground_truth == "Not Spam") & (predictions == "Spam"))
FN_mask = ((ground_truth == "Spam") & (predictions == "Not Spam"))
FP = np.sum(FP_mask)
FN = np.sum(FN_mask)
print("FP patients:", email_id[FP_mask])
print("FN patients:", email_id[FN_mask])
```

```python
study_hours = np.array([1, 2, 3, 5, 7, 8])
gaming_hours = np.array([6, 5, 4, 3, 2, 1])
actual_scores = np.array([35, 40, 50, 68, 82, 90])
weight1 = 8
weight2 = -3
bias = 45
X1 = study_hours
X2 = gaming_hours
predicted_score = X1 * weight1 + X2 * weight2 + bias
print(predicted_score)
predicted_score = np.clip(predicted_score,0 , 100)
print("predict:",predicted_score)

error = predicted_score - actual_scores
print("erro:",error)
absolute_error = np.abs(predicted_score-actual_scores)
print("abs: ",absolute_error)
MAE = np.mean(abs(predicted_score-actual_scores))
print("MAE: ",MAE)
```
